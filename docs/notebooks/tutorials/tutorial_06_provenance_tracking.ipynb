## Provenance tracking [draft tutorial]

When sharing or documenting results and workflows, it can be advantageous to keep a record of processing functions & settings that generated a specific dataset. 
`alphapepttools` supports tracking this information with its `provenance_tracker` flag.

```python
%load_ext autoreload
%autoreload 2

import anndata as ad
import numpy as np
import pandas as pd
import alphapepttools as apt
```

```python
# Small Example Dataset
numerical_data = np.array([[1, 0, 3], [4, 6, 9]])
sample_metadata = pd.DataFrame(
    {
        "obs_names": ["cell1", "cell2"],
        "age": [28, 29],
    }
).set_index("obs_names")
feature_metadata = pd.DataFrame({"var_names": ["gene1", "gene2", "gene3"]}).set_index("var_names")
```

### Provenance tracking of a standard workflow

```python
# Generate AnnData object
adata = ad.AnnData(
    X=numerical_data,
    obs=sample_metadata,
    var=feature_metadata,
)

# switch tracking on
apt.settings.track_provenance = True

# Apply log transformation
apt.pp.nanlog(adata)

# Apply imputation
apt.pp.impute_gaussian(adata)

# Visualize log
print(adata.uns["provenance"])
```
