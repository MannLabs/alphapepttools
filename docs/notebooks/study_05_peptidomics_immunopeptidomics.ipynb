```python
adata_precursor = io.read_psm_table(
    file_paths=proteomics_report,
    search_engine="diann",
    var_columns=["proteins", "genes", "sequence"],
)

adata_precursor = io.read_psm_table(
    file_paths=hla_report,
    search_engine="spectronaut_report",
    var_columns=["proteins", "genes", "sequence"],
)

adata_precursor = io.read_psm_table(
    file_paths=phospho_report,
    search_engine="spectronaut_report",
    var_columns=["proteins", "genes", "sequence"],
)
```
